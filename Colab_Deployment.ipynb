### Step 2: Start the Servers in the Background
This cell will start Python on port 8000, Java on port 8080, and React on port 5173 in the background of this Colab machine.

```python
import subprocess
import time
import os

os.chdir('/content/resume-app')

# 1. Start Python AI Engine
subprocess.Popen(["uvicorn", "api.main:app", "--port", "8000"])

# 2. Start Java Gateway
os.chdir('/content/resume-app/backend-java')
subprocess.Popen(["mvn", "spring-boot:run"])

# 3. Install NPM Packages & Start React
os.chdir('/content/resume-app/frontend-react')
!npm install
subprocess.Popen(["npm", "run", "dev", "--", "--host"])

os.chdir('/content/resume-app')
print("\n\n✅ All servers are booting up in the background! Wait 60 seconds for Java to compile.")
```
